```python
missing_mask = df[numeric_columns].isnull().all(axis=1)

missing_groups = (
    missing_mask.astype(int)
    .groupby((missing_mask != missing_mask.shift()).cumsum())
    .sum()
)

missing_groups[missing_groups > 0].describe()
```

```python
missing_starts = missing_mask & ~missing_mask.shift(fill_value=False)
missing_ends = missing_mask & ~missing_mask.shift(-1, fill_value=False)

start_indices = df.index[missing_starts]
end_indices = df.index[missing_ends]

gap_summary = pd.DataFrame({
    "start_index": start_indices,
    "end_index": end_indices
})

gap_summary["duration_minutes"] = (
    gap_summary["end_index"] - gap_summary["start_index"] + 1
)

gap_summary.sort_values(
    "duration_minutes",
    ascending=False
).head(10)
```

```python
df["DateTime"] = pd.to_datetime(
    df["Date"] + " " + df["Time"],
    dayfirst=True
)
```

```python
gap_summary["start_datetime"] = df.loc[
    gap_summary["start_index"], "DateTime"
].values

gap_summary["end_datetime"] = df.loc[
    gap_summary["end_index"], "DateTime"
].values
```

```python
small_gaps = gap_summary[gap_summary["duration_minutes"] <= 5]

large_gaps = gap_summary[gap_summary["duration_minutes"] > 5]

print("Small gaps:", len(small_gaps))
print("Large gaps:", len(large_gaps))
```

```python
print("Rows in small gaps:", small_gaps["duration_minutes"].sum())
print("Rows in large gaps:", large_gaps["duration_minutes"].sum())
```

```python
def categorize_gap(minutes):
    if minutes <= 5:
        return "Small"
    elif minutes <= 60:
        return "Medium"
    else:
        return "Long"

gap_summary["gap_category"] = gap_summary["duration_minutes"].apply(categorize_gap)

gap_summary["gap_category"].value_counts()
```
